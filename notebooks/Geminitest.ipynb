```python
def output_to_list(output):
    start = output.find("[")  # Find the starting bracket
    end = output.rfind("]")
    python_list = ast.literal_eval(output[start:end+1])
    print(python_list)

def output_to_dict(output):
    start = output.find("{")  # Find the starting bracket
    end = output.rfind("}")
    python_dict = ast.literal_eval(output[start:end+1])
    print(python_dict)

```
